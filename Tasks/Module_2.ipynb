```python
# Задание 2
number = int(input("Введите двузначное число: "))
left = number // 10
right = number % 10
new_number = right * 10 + left
print(new_number)
```

```python
# Задание 3
number = int(input("Введите многозначное число: "))
last_digits = number % 100
print('Последние 2 цифры числа:', last_digits)
```

```python
# Задание 4
number = int(input("Введите многозначное число: "))
tens = number // 10 % 10
print('Число в десятках: ', tens)
```

```python
# Задание 5
number = int(input("Введите трехзначное число: "))
hundreds = number // 100
tens = number // 10 % 10
units = number % 10
sum = hundreds + tens + units
print('Сумма цифр числа: ',sum)
```

```python
# Задание 6
number = float(input("Введите вещественное положительное число: "))
new_number = int(number * 10)
digit = new_number % 10
print('Первое число после запятой - ', digit)
```

```python
# Задание 7
year = int(input("Введите год: "))
century = ((year - 1) // 100) + 1
print('Век: ',century)
```

```python
# Задание 8
N = int(input("Введите день: "))
day_number = (4 + (N - 1)) % 7
print('Номер дня недели: ', day_number)
```

```python
# Задание 9
N = int(input("Введите кол-во минут после полуночи: "))
hours = N // 60
minuts = N % 60
print(hours, ':', minuts, sep="")
```

```python
# Задание 10
a = int(input('Введите кол-во детей в первом классе: '))
b = int(input('Введите кол-во детей во втором классе: '))
c = int(input('Введите кол-во детей в третьем классе: '))
p1 = (a - 1) // 2 + 1
p2 = (b - 1) // 2 + 1
p3 = (c - 1) // 2 + 1
print('Кол-во необходимых столов', p1 + p2 + p3, 'штук')
```
