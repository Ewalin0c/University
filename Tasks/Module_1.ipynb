```python
#Задание 1
a = int(input('Введите первое число: '))
b = int(input('Введите второе число: '))
c = int(input('Введите третье число: '))
print('Сумма чисел: ', a+b+c)
```

```python
#Задание 2
base = int(input('Введите значение основания треугольника: '))
height = int(input('Введите значение высоты треугольника: '))
print('Площадь треугольника: ', base * height / 2)
```

```python
#Задание 4
number = int(input('Введите число: '))
print('Следующее число для числа ', number, ': ', number + 1,  sep="")
print('Предыдущее число для числа ', number, ': ', number - 1,  sep="")
```

```python
#Задание 5
N = int(input("Введите количество студентов: "))
K = int(input("Введите количество яблок: "))
student_has = K // N
basket = K % N
print("Яблок у студентов: ", student_has, "; ", "в корзине: ", basket, sep="")
```

```python
#Задание 6
N = int(input("Введите число секунд после полуночи: "))
minutes = N // 60
hours = minutes // 60
print(hours, minutes)
```

```python
#Задание 7
h1 = int(input("Введите часы первой метки: "))
m1 = int(input("Введите минуты первой метки: "))
s1 = int(input("Введите секунды первой метки: "))
h2 = int(input("Введите часы второй метки: "))
m2 = int(input("Введите минуты второй метки: "))
s2 = int(input("Введите секунды второй метки: "))
time_difference = (h2 * 3600 + m2 * 60 + s2) - (h1 * 3600 + m1 * 60 + s1)
print("Разница: ", time_difference, "секунд")
```
